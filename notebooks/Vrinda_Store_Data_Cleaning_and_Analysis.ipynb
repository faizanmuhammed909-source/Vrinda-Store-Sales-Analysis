### Step 4: Data Cleaning and Preprocessing
Inspect unique values across all columns to detect anomalies, then clean the dataset by removing duplicate rows, correcting column data types (fixing the 'Qty' column issue), and standardizing categorical values (fixing the 'Gender' column issue).

#### Check Unique Values Across All Columns

```python
# Inspect unique values and categories across all columns to detect anomalies before cleaning
summary_data = []

for col in df.columns:
    n_unique = df[col].nunique()

    if n_unique < 15:
        unique_vals = list(df[col].unique())
    else:
        unique_vals = f"Continuous / ID column ({n_unique} unique values)"

    summary_data.append({
        'Column Name': col,
        'Unique Count': n_unique,
        'Unique Values / Summary': unique_vals
    })

# Render as a clean, Structured table
summary_df = pd.DataFrame(summary_data)
display(summary_df)
```

#### Check and Remove Duplicate Rows

```python
#Check for duplicate rows in the dataset
duplicate_count = df.duplicated().sum()
print(f"Total duplicate rows found: {duplicate_count}")

# Drop duplicates if they exist and reset the index
if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicate rows successfully removed.")
else:
    print("No duplicate rows to remove.")
```

#### Correct data Types (Qty Column)

```python
# Inspect unique values in 'Qty' to see what non-numeric entries are causing it to register as an object
print("Unique values in 'Qty':")
print(df['Qty'].unique())
```

```python
# Replace text values and convert the column to integer
df['Qty'] = df['Qty'].replace({'One':1, 'Two':2}).astype(int)

# Verify the updated data type and unique values
print("Updated Data Type:", df['Qty'].dtype)
print("Unique values:", df['Qty'].unique())
```

#### Standardize Gender Values (Gender Column)

```python
# Check unique values in Gender column before standardization
print("Unique values before fixing:", df['Gender'].unique())
```

```python
# Replace 'W' with 'Women' and 'M' with 'Men' 
df['Gender'] = df['Gender'].replace({'W': 'Women', 'M': 'Men'})

# Verify unique values after fixing
print("Unique values after fixing:", df['Gender'].unique())
```

### Step 5: Feature Engineering and Summary Statistics
Generate summary statistics for the dataset and create new features ('Month' and 'Age_Group') for deeper analysis

```python
# Generate summary statistics for dataset
df.describe()
```

```python
# Convet data to datetime and extract the month name
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month_name()

# Verify yhe new Month column
df['Month'].head()
```

```python
# Define bins and labels to categorize customer age into groups
bins = [0,19,35,60,100]
labels = ['Teenager', 'Young Adult', 'Adult', 'Senior']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

#Verify the new Age Group column along with Age
df[['Age', 'Age_Group']].head()
```

### Step 6: Exploratory Data Analysis & Group Aggregations
Analyze key metrics and customer purchasing patterns across engineering features ('Age_Group' and 'Month').

```python
# Group by Age Group to inspect total volume purchased per category 
age_group_summary = df.groupby('Age_Group')['Qty'].sum().reset_index()
display(age_group_summary)
```

```python
# Group by Month to inspect performance across time periods
monthly_summary =  df.groupby('Month')['Qty'].sum().reset_index()
display(monthly_summary)
```

```python
# Cross-tabulate Age Group and Month for demographic insights
pivot_summary = df.pivot_table(index='Age_Group', columns='Month', values='Qty', aggfunc='sum')
display(pivot_summary)
```
